#### Step 2: Load and Preview the Dataset

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
columns = ['Employment', 'JobSat', 'RemoteWork']
df[columns].isna().sum()
```

```python
# Employment - ✅ complete, no action needed
# RemoteWork - ⚠️ moderate missing (~16%) - safe to impute
# JobSat - ❗ very high missing (~55%) - is NOT safe to impute
# use conditional dropping for further analysis:
# df.dropna(subset=['JobSat'])

# Impute RemoteWork with most frequent value
df['RemoteWork'] = df['RemoteWork'].fillna(df['RemoteWork'].mode()[0])

df[['Employment', 'JobSat', 'RemoteWork']].isna().sum()
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
# Convert YearsCodePro to numeric
df['YearsCodePro_num'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

# create a clean Subset for JobSat analysis
df_exp_sat = df.dropna(subset=['YearsCodePro_num', 'JobSat']).copy()
```

```python
# experience ranges
bins = [0, 5, 10, 20, float('inf')]
labels = ['0–5', '5–10', '10–20', '>20']

df_exp_sat['ExperienceRange'] = pd.cut(
    df_exp_sat['YearsCodePro_num'],
    bins=bins,
    labels=labels,
    right=False
)
```

```python
median_jobsat = (
    df_exp_sat
    .groupby('ExperienceRange', observed=True)['JobSat']
    .median()
)
median_jobsat
```

```python
plt.figure(figsize=(8, 5))
median_jobsat.plot(kind='bar')
plt.title('Median Job Satisfaction by Experience Range')
plt.xlabel('Years of Professional Experience')
plt.ylabel('Median Job Satisfaction')
plt.xticks(rotation=0)
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
# Select relevant columns and drop missing values
df_lang = df[['Country', 'LanguageHaveWorkedWith']].dropna()

# Split and explode languages
df_lang['LanguageHaveWorkedWith'] = df_lang['LanguageHaveWorkedWith'].str.split(';')
df_lang = df_lang.explode('LanguageHaveWorkedWith')
df_lang.head()
```

```python
lang_region_counts = (
    df_lang
    .groupby(['Country', 'LanguageHaveWorkedWith'])
    .size()
    .reset_index(name='count')
)
lang_region_counts
```

```python
top_countries = df_lang['Country'].value_counts().head(10).index
top_languages = df_lang['LanguageHaveWorkedWith'].value_counts().head(10).index

lang_region_counts = lang_region_counts[
    lang_region_counts['Country'].isin(top_countries) &
    lang_region_counts['LanguageHaveWorkedWith'].isin(top_languages)
]

lang_region_counts
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(
    data=lang_region_counts,
    x='Country',
    y='count',
    hue='LanguageHaveWorkedWith'
)

plt.title('Top Programming Languages by Region')
plt.xlabel('Country')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
def group_employment(emp):
    if pd.isna(emp):
        return 'Other'
    
    emp_lower = emp.lower()

    if 'retired' in emp_lower:
        return 'Retired'
    if 'not employed' in emp_lower:
        return 'Not employed'
    if 'employed' in emp_lower:
        return 'Employed'
    
    return 'Other'

df['EmploymentGrouped'] = df['Employment'].apply(group_employment)
df['EmploymentGrouped'].value_counts()
```

```python
df.loc[df['EmploymentGrouped'] == 'Other', 'Employment'].value_counts()
```

```python
# redefine the employment grouping function
def group_employment(emp):
    if pd.isna(emp):
        return 'Other'
    
    emp_lower = emp.lower()

    if 'retired' in emp_lower:
        return 'Retired'
    if 'not employed' in emp_lower:
        return 'Not employed'
    if 'employed' in emp_lower:
        return 'Employed'
    if 'prefer not to say' in emp_lower:
        return 'Prefer not to say'
    if 'student' in emp_lower:
        return 'Student'
    return 'Other'

df['EmploymentGrouped'] = df['Employment'].apply(group_employment)
df['EmploymentGrouped'].value_counts()
```

```python
ed_emp_ct = pd.crosstab(
    df['EdLevel'],
    df['EmploymentGrouped']
)

ed_emp_ct
```

```python
ed_emp_pct = (
    pd.crosstab(
        df['EdLevel'],
        df['EmploymentGrouped'],
        normalize='index'
    ) * 100
).round(2)

ed_emp_pct
```

```python
ed_emp_pct.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6)
)

plt.title('Employment Type by Educational Background (%)')
plt.xlabel('Education Level')
plt.ylabel('Percentage')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Employment Type', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(
    ed_emp_pct,
    annot=True,
    fmt='.1f',
    cmap='Blues'
)

plt.title('Employment Type by Educational Background (%)')
plt.xlabel('Employment Type')
plt.ylabel('Education Level')
plt.show()
```
